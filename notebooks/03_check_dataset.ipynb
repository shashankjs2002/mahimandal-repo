# MahimandalSR dataset checks

These checks use the frozen manifest. They do not alter images, redraw the split, download data or run models.


```python
from pathlib import Path
import sys, subprocess, json
import pandas as pd
ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
MANIFEST = ROOT / "metadata/paper_release_manifest.csv"
OUTPUT = ROOT / "runs/dataset_checks"
OUTPUT.mkdir(parents=True, exist_ok=True)
```

```python
for script, report in (("audit_release_quality.py", "quality.json"),
                       ("audit_source_products.py", "source_overlap.json")):
    subprocess.run([sys.executable, str(ROOT / "dataset" / script), "--manifest", str(MANIFEST),
                    "--output", str(OUTPUT / report)], cwd=str(ROOT), check=True)
```

```python
display(pd.read_csv(OUTPUT / "quality_landscape_split.csv"))
print(json.dumps(json.loads((OUTPUT / "source_overlap.json").read_text())["per_split"], indent=2))
```
